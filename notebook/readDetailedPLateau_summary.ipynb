```python
# Import required libraries
# import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from plotly.subplots import make_subplots
import plotly.graph_objects as go

import plotly.express as px

```

```python
# load csv files into dataframe
csv_file_old = '/Users/evillz/Data/article/final_yey/w_felix/batch_analysis_results/detailed_plateau_data_old.csv'
csv_file_new = '/Users/evillz/Data/article/final_yey/w_felix/batch_analysis_results/detailed_plateau_data.csv'

df_old = pd.read_csv(csv_file_old)
df_new = pd.read_csv(csv_file_new)  
```

```python
# print columns names 
print(df_old.columns)
```

```python
# create histogram of desired key to compare both files
key = 'delta_avg'
plt.figure(figsize=(10, 6))
sns.histplot(df_old[key], color='blue', label='Old Data', kde=True, stat="density", bins=30, alpha=0.5)
sns.histplot(df_new[key], color='orange', label='New Data', kde=True, stat="density", bins=30, alpha=0.5)
plt.xlabel(key)
plt.ylabel('Density')
plt.title(f'Comparison of {key} Distributions')
plt.legend()
plt.show()
```

```python
# create for loop of keys to create subplot histograms
keys = ['delta_avg', 'delta_avg','rupture_slope','plateau_velocity_calc']
fig, axes = plt.subplots(len(keys), 1, figsize=(10, 6 * len(keys)))
for i, key in enumerate(keys):
    sns.histplot(df_old[key], color='blue', label='Old Data', kde=True, stat="density", bins=30, alpha=0.5, ax=axes[i])
    sns.histplot(df_new[key], color='orange', label='New Data', kde=True, stat="density", bins=30, alpha=0.5, ax=axes[i])
    axes[i].set_xlabel(key)
    axes[i].set_ylabel('Density')
    axes[i].set_title(f'Comparison of {key} Distributions')
    axes[i].legend()
plt.tight_layout()
plt.show()      
```
