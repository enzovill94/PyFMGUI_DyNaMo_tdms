```python
import matplotlib.pyplot as plt
import numpy as np  
import seaborn as sns
import pandas as pd
```

```python

fit_results = '/Users/evillz/Data/article/2025_07_01_THP1_phd/cell2/hertzfit_tingFit_cell2_ting_results.csv'
df_ting_results = pd.read_csv(fit_results)
ting = df_ting_results['ting_E0'].to_numpy()
hertz = df_ting_results['hertz_E'].to_numpy()

# plot me log scale histogram of hertz
# Define lower and upper limits for the histogram
lower_limit = -np.inf
upper_limit = np.inf

plt.figure(figsize=(10, 6))
plt.hist(
    hertz[(~np.isnan(hertz)) & (hertz >= lower_limit) & (hertz <= upper_limit)],
    bins=100,
    color='blue',
    alpha=0.7,
    edgecolor='black'
)
plt.xscale('log')
plt.yscale('log')
plt.title('Log-Log Histogram of Hertz Modulus')
plt.xlabel('Hertz Modulus (Pa)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75, which='both')
plt.show()
```

```python
# Filter hertz within lower and upper limits and remove NaNs
E0_filtered = hertz[(~np.isnan(hertz)) & (hertz >= lower_limit) & (hertz <= upper_limit)]

plt.figure(figsize=(10, 6))
sns.histplot(E0_filtered, bins=10, kde=True, color='blue', stat='density')
plt.xscale('log')
plt.title('Distribution of E_0 (Filtered)')
plt.xlabel('E_0 (Pa)')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Filter ting within lower and upper limits and remove NaNs
ting_filtered = ting[(~np.isnan(ting)) & (ting >= lower_limit) & (ting <= upper_limit)]

plt.figure(figsize=(10, 6))
sns.histplot(ting_filtered, bins=100, kde=True, color='blue', stat='density')
plt.xscale('log')
plt.title('Distribution of Ting E0 (Filtered)')
plt.xlabel('Ting E0 (Pa)')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
