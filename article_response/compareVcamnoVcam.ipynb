```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# create a dataframe with the data from the tdms files
wo_vcam = '/Volumes/SSK_Drive/thp1_last_experiment_2026/2026_29_05_THP1_final/noVCAM/tether_session_20260602_215237.csv'
w_vcam = '/Volumes/SSK_Drive/thp1_last_experiment_2026/2026_29_05_THP1_final/moreVcam/tether_session_20260602_215916.csv'
```

```python
# show first 5 rows of the dataframes and columns
wo_vcam_df = pd.read_csv(wo_vcam)
w_vcam_df = pd.read_csv(w_vcam) 

# print(wo_vcam_df.head())
print(wo_vcam_df.columns)
# print ("____________________________________________________________")
# print(w_vcam_df.head())
print(w_vcam_df.columns)
```

```python
# plot vs total files of bool_good_curve of both dataframes boolean 1 or 0  
# 0 meaning no tether event and 1 meaning a tether event    
plt.figure(figsize=(10,5))  
sns.countplot(x='bool_good_curve', data=wo_vcam_df, color='blue', label='No VCAM')
sns.countplot(x='bool_good_curve', data=w_vcam_df, color='orange', label='With VCAM', alpha=0.5)
plt.legend()
plt.title('Comparison of Tether Events with and without VCAM')
plt.xlabel('Tether Event (0 = No Event, 1 = Adhesion Events)')
# show only 0 and 1 on x-axis   
# plt.xticks([0, 1])
plt.ylabel('Count')
plt.show()
```
